```python
# FI 422 - Homework 4: Personal Finance Tracker
# Exercises 1-4: Record expenses, summarize spending, find insights,
# and compare spending with a weekly budget.

# Exercise 1: Record expenses in a list of dictionaries.
expenses = []

print("Personal Finance Tracker")
print("Enter at least 6 expenses across at least 3 categories.")

while True:
    item = input("Enter item name (or 'done' to stop): ").strip()

    if item.lower() == "done":
        categories = []
        for expense in expenses:
            if expense["category"] not in categories:
                categories.append(expense["category"])

        if len(expenses) < 6 or len(categories) < 3:
            print("Please enter at least 6 expenses and 3 different categories before finishing.")
            continue
        break

    if item == "":
        print("Please enter an item name.")
        continue

    category = input("Enter category: ").strip().lower()
    if category == "":
        print("Please enter a category. This expense was not saved.")
        continue

    amount = float(input("Enter amount: "))
    expenses.append({"item": item, "category": category, "amount": amount})

print("\nYour expenses:", expenses)


# Exercise 2: Sum spending in each category using a dictionary and a loop.
def summarize_by_category(expenses):
    summary = {}
    for expense in expenses:
        category = expense["category"]
        amount = expense["amount"]
        if category not in summary:
            summary[category] = 0
        summary[category] = summary[category] + amount
    return summary


summary = summarize_by_category(expenses)
print("\nCategory summary:")
for category in summary:
    print(category + ": $" + format(summary[category], ".2f"))


# Exercise 3: Find the largest category and spending statistics.
def largest_category(summary_dict):
    largest = None
    for category in summary_dict:
        if largest is None or summary_dict[category] > summary_dict[largest]:
            largest = category
    return largest


def spending_stats(summary_dict):
    total = 0
    for category in summary_dict:
        total = total + summary_dict[category]
    average = total / len(summary_dict)
    return total, average


biggest = largest_category(summary)
total, average = spending_stats(summary)
print("\nLargest spending category:", biggest)
print("Total spent: $" + format(total, ".2f"))
print("Average per category: $" + format(average, ".2f"))


# Exercise 4: Compare weekly spending to the user's budget.
budget = float(input("\nEnter your weekly budget: "))
if total <= budget:
    print("You are within your budget!")
    print("Amount remaining: $" + format(budget - total, ".2f"))
else:
    print("You went over your budget by $" + format(total - budget, ".2f") + ".")
```
